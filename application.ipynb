```python
# 이 셀은 커널 재시작 후 한 번만 실행. 다시 실행하면 pending 걸릴 수 있음
from langchain_ollama import ChatOllama
```

```python
# 위에서 임포트한 뒤, 이 셀도 한 번만 실행
llm = ChatOllama(model="llama3.2:1b")
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts  import PromptTemplate

# 음식 이름 찾기
output_parser = StrOutputParser()
# 줄바꿈도 다 토큰임 그래서 웬만하면(가독성 생각) 줄바꿈 쓰지 않는게 좋음
# 논문에 따르면 존댓말, please, thank you 등 조금 더 친절한 표현을 쓰는게 좋음 => 더 좋은 답변 받을 수 있음
country_food_prompt = PromptTemplate(
    template="""What is the most famous food in {country}? Please return the name of the food only""",
    input_variables=["country"]
)

country_food_chain = country_food_prompt | llm | output_parser

country_food_chain.invoke({"country": "France"})
```

```python
from langchain_core.prompts import ChatPromptTemplate

# food_recipe_prompt = PromptTemplate(
#     template="""What is the recipe for {food}? Return the recipe only""",
#     input_variables=["food"]
# )

food_recipe_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that can answer questions about recipes."),
    ("human", "What is the recipe for {food}?")
])

food_recipe_chain = food_recipe_prompt | llm | output_parser

food_recipe_chain.invoke({"food": "pizza"})
```

```python
final_chain = {"food": country_food_chain} | food_recipe_chain

final_chain.invoke({"country": "France"})
# 인풋보다 아웃풋이 훨씬 비쌈 그래서 돈이 더 나오니까 가급적이면 간결한, 딱 우리가 원하는 아웃풋을 만들도록 해야함
```
